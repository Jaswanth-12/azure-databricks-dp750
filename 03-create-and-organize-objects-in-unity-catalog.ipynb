### Task 2.3 — Create the `enrollments` table with foreign keys

Create `edu_dev_gjs.silver.enrollments` to record which students are enrolled in which courses. This table must reference both `students` and `courses` via foreign key constraints.

| Column | Type | Description |
|--------|------|-------------|
| `enrollment_id` | BIGINT | Unique enrollment record (primary key) |
| `student_id` | BIGINT | References `students.student_id` |
| `course_id` | BIGINT | References `courses.course_id` |
| `semester` | STRING | Academic semester (e.g., `Spring 2023`) |
| `grade` | DECIMAL(4,2) | Numerical grade on a 0.0–10.0 scale |

> 🤖 **Genie Code tip:** Ask:
> *"How do I add foreign key constraints referencing other tables when creating a table in Databricks Unity Catalog?"*

**Hint:** Use `CONSTRAINT <name> FOREIGN KEY (<col>) REFERENCES <catalog>.<schema>.<table>(<col>)` for each foreign key.

````python
%sql
-- TODO: Create edu_dev_gjs.silver.enrollments with primary key and two foreign key constraints
CREATE TABLE IF NOT EXISTS edu_dev_gjs.silver.enrollments(
    enrollment_id BIGINT PRIMARY KEY,
    student_id BIGINT REFERENCES edu_dev_gjs.silver.students(student_id),
    course_id BIGINT REFERENCES edu_dev_gjs.silver.courses(course_id),
    semester STRING,
    grade DECIMAL(3,2)
)
````

## Exercise 3: Create Views

Views provide a virtual layer over your tables, simplifying access to joined data and pre-computing expensive aggregations. In this exercise, you create a **standard view** for combined enrollment records and a **materialized view** in the `gold` schema to serve the Lakeside University analytics team with pre-aggregated department statistics.

### Task 3.1 — Create a standard view

Create a view `edu_dev_gjs.silver.vw_student_enrollments` that presents a combined, human-readable view of student enrollment records. It should include:

- Student full name (concatenate `first_name` and `last_name` with a space between them)
- Student `email`
- `course_name`
- `department`
- `semester`
- `grade`

> 🤖 **Genie Code tip:** Ask:
> *"Write a SQL view in Azure Databricks that joins three tables: students, courses, and enrollments"*

**Hint:** Use `CONCAT(first_name, ' ', last_name)` or the `||` operator to combine names. You need to join all three silver tables.

````python
%sql
-- TODO: Create the edu_dev_gjs.silver.vw_student_enrollments view
-- Join students, courses, and enrollments; include full_name, email, course_name, department, semester, grade

CREATE VIEW IF NOT EXISTS edu_dev_gjs.silver.vw_student_enrollments AS
SELECT
    CONCAT(s.first_name, ' ', s.last_name) AS full_name,
    s.email,
    c.course_name,
    c.department,
    e.semester,
    e.grade
FROM
    edu_dev_gjs.silver.students s
JOIN
    edu_dev_gjs.silver.courses c
ON
    s.program = c.department
JOIN
    edu_dev_gjs.silver.enrollments e
ON
    s.student_id = e.student_id
````

````python
%sql
SELECT *, dense_rank() OVER(PARTITION BY full_name ORDER BY course_name) as d_rank
FROM edu_dev_gjs.silver.vw_student_enrollments;
````

### Task 3.2 — Create a materialized view for department statistics

Create a materialized view `edu_dev_gjs.gold.vw_department_enrollment_stats` that pre-computes the following per academic department:

| Output column | Description |
|--------------|-------------|
| `department` | Academic department name |
| `total_enrollments` | Total number of enrollment records |
| `avg_grade` | Average grade, rounded to 2 decimal places |
| `distinct_students` | Number of unique students enrolled |

> 🤖 **Genie Code tip:** Ask:
> *"What is the syntax for creating a materialized view in Azure Databricks that aggregates data from two joined tables?"*

**Hint:** Use `CREATE MATERIALIZED VIEW`, join `enrollments` with `courses`, and use `COUNT()`, `ROUND(AVG(...), 2)`, and `COUNT(DISTINCT ...)`.

> [!NOTE]
> Materialized views require a SQL warehouse to run. If this cell fails, check the compute selector at the top of the notebook and switch from **Serverless** compute to a **Serverless SQL warehouse**.

````python
%sql
-- TODO: Create the edu_dev_gjs.gold.vw_department_enrollment_stats materialized view
CREATE MATERIALIZED VIEW IF NOT EXISTS edu_dev_gjs.gold.vw_department_enrollment_stats AS
SELECT
    department,
    COUNT(*) AS total_enrollments,
    ROUND(AVG(grade),2) AS avg_grade,
    COUNT(DISTINCT full_name) AS distinct_students
FROM
    edu_dev_gjs.silver.vw_student_enrollments
GROUP BY
    department
````

## Exercise 5: Create a Reusable SQL Function

Unity Catalog allows you to create **governed, reusable functions** that encapsulate business logic. Instead of repeating grade conversion logic in every query, you'll write it once as a SQL function and call it from any notebook or query in the workspace.

### Task 5.1 — Create a grade classification function

Create a SQL scalar function `edu_dev_gjs.silver.get_grade_classification` that accepts a `grade DECIMAL(4,2)` and returns a `STRING` letter classification based on the scale used at Lakeside University:

| Grade range | Classification |
|-------------|----------------|
| ≥ 8.5 | `'A'` |
| ≥ 7.0 | `'B'` |
| ≥ 5.5 | `'C'` |
| ≥ 4.0 | `'D'` |
| < 4.0 | `'F'` |

> 🤖 **Genie Code tip:** Ask:
> *"How do I create a SQL scalar function in Unity Catalog that uses a CASE expression to return a letter grade based on a numeric score?"*

**Hint:** Use `CREATE FUNCTION ... RETURNS STRING RETURN CASE WHEN ... END`.

````python
%sql
-- TODO: Create the edu_dev_gjs.silver.get_grade_classification function
CREATE OR REPLACE FUNCTION get_grade_classification(grade DECIMAL(4,2))
RETURNS STRING
RETURN CASE
  WHEN grade >= 8.5 THEN 'A'
  WHEN grade >= 7.0 THEN 'B'
  WHEN grade >= 5.5 THEN 'C'
  WHEN grade >= 4.0 THEN 'D'
  ELSE 'F'
END
````

### Task 5.2 — Test the function

Query `edu_dev_gjs.silver.enrollments` and apply your new function to produce a `grade_classification` column for each row. Include: `enrollment_id`, `student_id`, `course_id`, `semester`, `grade`, and `grade_classification`.

> 🤖 **Genie Code tip:** Ask:
> *"How do I call a Unity Catalog user-defined function in a SELECT statement in Azure Databricks?"*

````python
%sql
-- TODO: Query edu_dev_gjs.silver.enrollments and call get_grade_classification on the grade column
SELECT
    enrollment_id,
    student_id,
    course_id,
    semester,
    grade,
    get_grade_classification(grade) AS grade_classification
FROM edu_dev_gjs.silver.enrollments;
    
````
